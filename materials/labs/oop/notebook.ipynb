```python
# membuat aplikasi menu restoran (OOP)
# atribute
# - listOrder -> berupa list -> untuk menyimpan pesanan
# method
# - mencatat pesanan
# - menampilkan pesanan

# class -> PascalCase
# function -> camelCase
class App():
    def __init__(self):
        self.items = []
    
    def takeOrder(self, nama_makanan):
        # mencatat pesanan
        self.items.append(nama_makanan)

    def showOrders(self):
        # menampilkan pesanan
        return self.items


# membuat object App
app = App()

while True:
    print("selamat datang di resto fox")
    print("silakan pilih menu berikut")
    print("1. tambah pesanan")
    print("2. tampilkan pesanan")
    print("3. exit")
    result = input("masukkan angka menu")

    if result == '1':
        makanan = input("masukkan nama makanan")
        app.takeOrder(makanan)
    elif result == '2':
        print(app.showOrders())
    elif result == '3':
        print("program selesai")
        break
    else:
        print("salah menu! coba lagi")
```

```python
# inheritance
class Parent:
    def __init__(self):
        self.hairColor = "blonde"

    def makeSound(self):
        return "quack quack"

class Child(Parent):
    # overriding
    def makeSound(self):
        return "meow meow"


parent = Parent()
child = Child()

print(child.makeSound())
```
